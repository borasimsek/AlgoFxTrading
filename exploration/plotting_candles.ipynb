```python
import pandas as pd
import plotly.graph_objects as go
import datetime as dt
```

```python
pair = 'EUR_USD'
granularity = 'H4'
df = pd.read_pickle(f'../data/{pair}_{granularity}.pkl')
df['sTime'] = [dt.datetime.strftime(i, "s%Y-%m-%d %H:%M") for i in df['time']]
```

```python
df.head()
```

```python
df_plot = df.iloc[:100]
df_plot.head()
```

```python
fig = go.Figure()
fig.add_trace(go.Candlestick
            (x=df_plot['sTime'],
                open=df_plot['mid_o'],
                high=df_plot['mid_h'],
                low=df_plot['mid_l'],
                close=df_plot['mid_c'],
                name='Candlesticks'))

# Note that there are missing data (due to weekends)
```

```python
fig = go.Figure()
fig.add_trace(go.Candlestick(
                x=df_plot['sTime'],
                open=df_plot['mid_o'],
                high=df_plot['mid_h'],
                low=df_plot['mid_l'],
                close=df_plot['mid_c'],
                name='Candlesticks'),
                )
fig.update_yaxes(
    gridcolor='rgb(15,15,15)',
)

fig.update_xaxes(
    gridcolor='rgb(15,15,15)',
    rangeslider_visible=False,
    nticks = 10,
)
fig.update_layout(
    width=1500,
    height=400,
    margin = dict(l = 10, r = 10, t = 10, b = 10),
    paper_bgcolor='#1f2630',
    plot_bgcolor='#1f2630',
    font = dict(size = 8, color='white'),
)

fig.show()
```
